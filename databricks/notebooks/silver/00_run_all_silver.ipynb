`05_silver_products`, `06_silver_order_reviews`, `07_silver_order_payments`, `08_silver_sellers`, `09_silver_checks`. Il s'arrête au premier échec.

**Limite :** les lignes `✅`/`❌` de contrôle de chaque notebook ne s'affichent pas ici.

```python
notebooks = [
    "02_silver_orders",
    "03_silver_customers",
    "04_silver_order_items",
    "05_silver_products",
    "06_silver_order_reviews",   # a besoin de silver_orders : l'ordre compte
    "07_silver_order_payments",
    "08_silver_sellers",
    "09_silver_checks",
]

for nom in notebooks:
    print(f"▶️  {nom} ...")
    try:
        dbutils.notebook.run(f"./{nom}", 1200)   # 1200 s maximum par notebook
        print(f"✅ {nom} terminé")
    except Exception as e:
        print(f"❌ {nom} a échoué : {str(e)[:400]}")
        print("Arrêt : ouvre ce notebook seul pour voir l'erreur complète.")
        break
```

## Résumé : combien de lignes dans chaque table Silver ?

```python
attendus = {
    "silver_orders": 99441,
    "silver_customers": 99441,
    "silver_order_items": 112650,
    "silver_products": 32951,
    "silver_order_reviews": 99224,
    "silver_order_payments": 103886,
    "silver_sellers": 3095,
}
for table, attendu in attendus.items():
    try:
        n = spark.sql(f"SELECT COUNT(*) FROM workspace.silver.{table}").collect()[0][0]
        print(("✅ OK     " if n == attendu else "❌ ÉCART  ") + f"{table} : obtenu = {n} | attendu = {attendu}")
    except Exception as e:
        print(f"❌ {table} introuvable : {str(e)[:150]}")
```
