# Gold : `dim_products`

**Entrée :** `workspace.silver.products`
**Sortie :** `workspace.gold.dim_products` (une ligne par produit, clé : `product_id`)

## Ce qui change de Silver vers Gold
| Silver | Gold |
|---|---|
| Toutes les colonnes de la source, plus les indicateurs | Seulement ce qui sert au dashboard et à l'agent |
| Noms hérités de la source | Noms clairs : `category`, `category_pt`, `weight_g`... |
| Indicateurs `flag_...` | Un seul booléen lisible : `category_unknown` |

Les colonnes en anglais facilitent la tâche de l'agent text-to-SQL.

```sql
CREATE OR REPLACE TABLE workspace.gold.dim_products AS
SELECT
    product_id,
    COALESCE(product_category_name_english, product_category_name) AS category,
    product_category_name         AS category_pt,
    flag_category_missing         AS category_unknown,
    product_weight_g              AS weight_g,
    product_length_cm             AS length_cm,
    product_height_cm             AS height_cm,
    product_width_cm              AS width_cm,
    product_photos_qty            AS photo_count
FROM workspace.silver.silver_products
```

## Vérification
| Mesure | Attendu |
|---|---:|
| `produits`, `produits_distincts` | 32 951 |
| `categories` | nombre de catégories, `unknown` compris |

```sql
SELECT
    COUNT(*) AS produits,
    COUNT(DISTINCT product_id) AS produits_distincts,
    COUNT(DISTINCT category) AS categories,
    COUNT(*) FILTER (WHERE category_unknown) AS categorie_inconnue
FROM workspace.gold.dim_products
```
